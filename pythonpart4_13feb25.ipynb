```python
def additiion(num1, num2):
  result = num1 + num2
  print("Result inside Function: ",result)

print("Before calling Function")
additiion(10,20)
print("After calling Function")
```

```python
#if number of arguments are unknown
def additiion(*num):
  result = 0
  for i in num:
    result = result + i
  print("Result inside Function: ",result)
print("Addition of three number")
additiion(10,20,30)
print("Addition of four number")
additiion(10,20,30,40)
print("Addition of five number")
additiion(10,20,30,40,50)
print("Addition of single number")
additiion(10)
```
