**class**\
class is a combination of related data

object is the instance of the class

```python
class Employee:
  empId = 101
  fname = "Raju"
  lname = "Doe"
  address = "Bangalore"

emp1 = Employee() #emp1 is the object of the Employee class
print("Employee ID: ",emp1.empId)
print("Employee First Name: ",emp1.fname)
print("Employee Last Name: ",emp1.lname)
print("Employee Address: ",emp1.address)
```

built in class attributes

```python
class Employee :
  def __init__(self, name="Raju", id = 101, address = "Bangalore"):
    self.ename = name
    self.id = id
    self.address = address

emp = Employee("John", 201,"Hyderabad")
print("Employee Name:", emp.ename)
print("Employee ID:",emp.id)
print("Employee Address:",emp.address)


```
