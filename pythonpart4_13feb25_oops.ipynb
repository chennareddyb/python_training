```python
def additiion(num1, num2):
  result = num1 + num2
  print("Result inside Function: ",result)

print("Before calling Function")
additiion(10,20)
print("After calling Function")
```

```python
#if number of arguments are unknown and it will take in tuple format
def additiion(*num):
  result = 0
  for i in num:
    result = result + i
  print("Result inside Function: ",result)
print("Addition of three number")
additiion(10,20,30)
print("Addition of four number")
additiion(10,20,30,40)
print("Addition of five number")
additiion(10,20,30,40,50)
print("Addition of single number")
additiion(10)
```

```python
def empName(*name):
  print("Name of third employee is :", name[2])
empName("chenna","Raju","Suman")
```

```python
def empDetails(**emp): #if we want details in dictionary format then we use **
  print("ID of employee",emp["id"])
  print("Name of employee",emp["fname"],emp["lname"])
  print("Employee: ",emp)
empDetails(fname="Raju",lname="singh", id=101)
```

```python
# setting default parameter
def projectData(pname = "Internal"):
  print("Project Name: ",pname)
projectData("BMW")
projectData()
projectData("Citi Bank")
```

```python
def location(loc): #passing list of data
  for officeloc in loc:
    print(officeloc)

location(["Chennai","Mumbai","Delhi"])
```

```python
#Lambda functions (It means no name to the functions)
#lambda arguments:expression
x = lambda num1,num2: num1 + num2
print("Addition of two numbers: ",x(10,20))
```

```python
x = lambda num: num*13
print(x(7))
```

```python
calc = lambda num1, num2, num3: num1 + num2 - num3
print(calc(20,20,30))
```

```python
name = "John Doe singh"
upper = lambda param: param.upper()
print(upper(name))
```

```python
maxValue = lambda num1, num2 : num1 if(num1>num2) else num2
print("Max Value is ",maxValue(10,20))
```

```python
# return value from function
def addToValues(num1, num2):
  return num1 + num2
print("Addition of two numbers: ",addToValues(10,20))#print the output

result = addToValues(10,20) # store the output into result variable
print("Addition of two numbers store in result: ",result)
```
