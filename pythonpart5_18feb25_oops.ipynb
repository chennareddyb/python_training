```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age
  def displayData(self):
    print("Pesrion Name: ",self.name, " Person Age:", self.age)
    print("Person __doc__: ",Person.__doc__)
    print("Person class name __name__ : ",Person.__name__)
    print("Person __module__: ",Person.__module__)
    print("Person __bases__: ",Person.__bases__)
    print("Person __dict__: ",Person.__dict__)

p1 = Person("John", 36)

p1.displayData()
```

```python
class Employee:
  empCount = 0
  def __init__(self, name, id):
    self.name = name
    self.id = id
    Employee.empCount += 1
  def displayCount():
    print("Total Employee %d" % Employee.empCount)
  def displayEmployee(self):
    print("Name : ", self.name,  ", ID: ", self.id)
  empCounter = staticmethod(displayCount)

emp1 = Employee("Zaraa", 2000)
emp2 = Employee("Mannii", 5000)
emp1.displayEmployee()
emp2.displayEmployee()
Employee.empCounter()
```

```python
#__str__ this is used to convert the object into string and we have to return the class object in the form of String
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age
  def __str__(self):
    return f"{self.name}-{self.age}"

p1 = Person("Johnn", 36)

print(p1)
```

delattr will delete attribute, in below case we deleted id attribute so if we try to use again in display method then it will fail


```python
class Employee:
  def __init__(self, name, id, address):
    self.name = name
    self.id = id
    self.address = address
  def __del__(self):
    print("Employee deleted")
  def display(self):
    print("Name : ", self.name,  ", ID: ", self.id)

emp1 = Employee("Zaraa", 2000, "Delhi")
print(getattr(emp1, 'name'))
print(setattr(emp1, 'address', 'Bengalore'))
print(getattr(emp1, 'address'))
print(hasattr(emp1, 'id'))
delattr(emp1, 'id')
print(hasattr(emp1, 'id'))
emp1.display()

```
