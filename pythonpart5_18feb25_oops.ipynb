**class**\
class is a combination of related data

object is the instance of the class

__doc__\
__name__\
__module__\
__bases__\
__dict__

```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age
  def displayData(self):
    print("Pesrion Name: ",self.name, " Person Age:", self.age)
    print("Person __doc__: ",Person.__doc__)
    print("Person class name __name__ : ",Person.__name__)
    print("Person __module__: ",Person.__module__)
    print("Person __bases__: ",Person.__bases__)
    print("Person __dict__: ",Person.__dict__)

p1 = Person("John", 36)

p1.displayData()
```

classmethod() is a predefined method


```python
#counter
class Employee:
  empCount = 0
  def __init__(self, name, salary):
    self.name = name
    self.salary = salary
    Employee.empCount += 1
  def displayCount(self):
    print("Total Employee %d" % Employee.empCount)
  employeeCount = classmethod(displayCount)
  def displayEmployee(self):
    print("Name : ", self.name,  ", Salary: ", self.salary)


emp1 = Employee("Zaraa", 2000)
emp2 = Employee("Mannii", 5000)
emp1.displayEmployee()
emp2.displayEmployee()
emp1.displayCount()
Employee.employeeCount()
```

```python
class Employee:
  empCount = 0
  def __init__(self, name, id):
    self.name = name
    self.id = id
    Employee.empCount += 1
  def displayCount():
    print("Total Employee %d" % Employee.empCount)
  def displayEmployee(self):
    print("Name : ", self.name,  ", ID: ", self.id)
  empCounter = staticmethod(displayCount)

emp1 = Employee("Zaraa", 2000)
emp2 = Employee("Mannii", 5000)
emp1.displayEmployee()
emp2.displayEmployee()
Employee.empCounter()
```

classmethod and staticmethod are serving the same purpose
