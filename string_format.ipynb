```python
result = 100/777

print(result)
```

```python
print('The result is {}'.format(result))
```

```python
print('The result is {r}'.format(r=result))
```

```python
print('The result is {r:1.3f}'.format(r=result))
# {value:width.precision f} if we give width more it will ad empty space before the calue
```

```python
result = 105.123456
print('The result is {r:1.3f}'.format(r=result))
```

r - read only

w - write only

a - append only

r+ is reading and writing

w+ writing and reading

```python
%%writefile new_file.txt
ONE ON FIRST
TWO ON SECOND
THREE ON THIRD
Fouur on Fouur

```

```python
with open('new_file.txt', mode='r') as f:
    print(f.read())
```

```python
with open('new_file.txt', mode='a') as f:
    f.write('FIVE ON FIFTH')

with open('new_file.txt', mode='r') as f:
    print(f.read())
```
