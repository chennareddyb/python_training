```python
print('This is first python Test')
```

```python
print('This is {} {} {}'.format('first', 'python', 'Test'))
```

```python
print('This is {0} {2} {1}'.format('first', 'python', 'Test'))
```

```python
print('This is {f} {p} {t}'.format(f='first', p='python', t='Test'))
```

```python
result = 100/777

print(result)
```

```python
print('The result is {}'.format(result))
```

```python
print('The result is {r}'.format(r=result))
```

```python
print('The result is {r:1.3f}'.format(r=result))
# {value:width.precision f} if we give width more it will ad empty space before the calue
```

```python
result = 105.123456
print('The result is {r:1.3f}'.format(r=result))
```
