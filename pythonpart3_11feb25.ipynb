**Dictionary**

denoted as dict

It stores values in key: value pairs\
maintanins the insersion order\
key cannot be duplicate



```python
myDict = {"name":"Chenna", "age":25, "city":"Hyderabad"}
print(myDict)
print(type(myDict))
print(len(myDict))
print(myDict["name"])
```

```python
myKeys = myDict.keys()
print(myKeys)
print(type(myKeys))
myValues = myDict.values()
print(myValues)
print(type(myValues))
```

```python
for(x,y) in myDict.items():
  print(x,y)

for x in myDict.keys():
  print(x)

for y in myDict.values():
  print(y)
```

```python
#create dictionary using constructor
myDict1 = dict(name="Chenna", age=25, city="Hyderabad")
print(myDict1)
```

```python
city = myDict1.get("city")
print(city)
```

```python
myDict1["age"] = 26
print(myDict1)

myDict1.update({"age":27})
print(myDict1)
```

```python
myDict1.update({"country":"India"})
print(myDict1)
```

```python

myDict1.pop("country")
print(myDict1)
```

```python
#convert to tuple
myData = myDict1.items()
print(myData)
print(type(myData))
```

```python
#delete last inserted value
myDict1.popitem()
print(myDict1)
myDict1.clear()
print(myDict1)
```
