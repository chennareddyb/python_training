```python
nameList = ["Abc", "Bcd", "Cde", "Def", "Efg", "Fgh", "Hij"]
print(nameList)
```

```python
if "Cde" in nameList:
  print("Yes it is present")
```

```python
print(nameList[2:6])
```

```python
tmpList = nameList
tmpList[2:5]= ["Xyx","Yyz"]
print(tmpList)
print(nameList)
```

```python
#insert new value in the list
nameList.insert(2,"Hendry")
print(nameList)
nameList.append("Chenna")
print(nameList)
```

```python

list2 = [1,2,3,4]
print(list2)
nameList.extend(list2)
print(nameList)
```

```python
#Deleting data from the list
myList = ["Abc","Bcd","Efg","Abc","Def","abc"]
print(myList)
myList.remove("Abc")
print(myList)
```

```python
myList = ["Abc","Bcd","Efg","Abc","Def","abc"]
print(myList)
myList.pop(1) #remove the index value from the list
print(myList)
myList.pop() # remove t last value as list works as LIFO (last in first out)
print(myList)
```

```python
myList = ["Abc","Bcd","Efg","Abc","Def","abc"]
print(myList)
del myList[0] # delete teh given index
print(myList)
```

```python
myList = ["Abc","Bcd","Efg","Abc","Def","abc"]
print(myList)
# delete complete list
del myList
print(myList)
```
