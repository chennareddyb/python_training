```python
#Sorting the list
list1 = ["Raju","Hari","Baba","Ali","kohli"]
print(list1)
list1.sort()
print(list1)# By default ascending
list1.sort(reverse=True)
print(list1) # sort in decending order
```

```python
#Display in reverse order
list1 = ["Raju","Hari","Baba","Ali","kohli"]
print(list1)
list1.reverse()
print(list1)
list2 = list1.copy() # copy the list data into new variable
print(list2)
```
